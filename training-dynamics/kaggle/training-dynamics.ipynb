validation metric on a held-out split, and recorded what the interval cost in
wall-clock minutes.

This notebook **does not train anything**. It analyses the records the training queue
wrote while the runs happened. Every number below is computed from those records in

## The stopping rule

A chunk's gain is judged **relative** to where the metric already stood, not as an
absolute delta. The threshold is 0.2%: on a PSNR of 27 dB that is 0.054 dB, well below
what is visible in an output image.

Relative matters. An absolute threshold treats +0.001 dB at 27 dB as the same event as
+0.001 dB at 12 dB. And in an earlier version any new best reset the counter, so a gain
of +0.0013 dB bought another four hours of training.

```python
FLOOR = 0.2  # percent

for c in CHUNKS:
    c["gain_pct"] = (c["after"] - c["before"]) / abs(c["before"]) * 100

under = [c for c in CHUNKS if c["gain_pct"] < FLOOR]
total_h = sum(c["minutes"] for c in CHUNKS) / 60
under_h = sum(c["minutes"] for c in under) / 60

print(f"intervals under the {FLOOR}% floor : {len(under)} of {len(CHUNKS)}")
print(f"hours spent on them            : {under_h:.1f} of {total_h:.1f}")
print(f"share of measured GPU time     : {under_h / total_h * 100:.0f}%")
```

## What the expensive intervals bought

Sorted by cost. These are the intervals where the most time produced the least
movement, and two of them moved the metric *backwards* - meaning the published weights
for those models come from an earlier point in the run.

```python
worst = sorted(under, key=lambda c: -c["minutes"])[:10]
print(f"{'model':22}{'ended at':>12}{'minutes':>9}{'gain':>10}")
for c in worst:
    at = f"{c['at']} {c['unit']}"
    print(f"{c['model']:22}{at:>12}{c['minutes']:>9.0f}{c['gain_pct']:>+9.3f}%")
```

## Cost against gain

One dot per interval. The dashed line is the 0.2% floor. Points are coloured by which
side of it they fall on, and the four costliest no-gain intervals are labelled
directly, so identity never depends on colour alone.

```python
import matplotlib.pyplot as plt

BLUE, ORANGE, MUTED = "#2a78d6", "#eb6834", "#7a7873"
fig, ax = plt.subplots(figsize=(10, 4.6), dpi=140)

series = ((lambda c: c["gain_pct"] >= FLOOR, BLUE, f"gained more than {FLOOR}%"),
          (lambda c: c["gain_pct"] < FLOOR, ORANGE, f"under the {FLOOR}% floor"))
for keep, col, lab in series:
    pts = [c for c in CHUNKS if keep(c)]
    ax.scatter([c["minutes"] for c in pts], [c["gain_pct"] for c in pts],
               s=46, c=col, edgecolors="white", linewidths=1.2, label=lab, zorder=3)

ax.axhline(0, color=MUTED, lw=1, zorder=1)
ax.axhline(FLOOR, color=ORANGE, lw=1.6, ls=(0, (5, 4)), zorder=2)
ax.annotate(f"{FLOOR}% floor", xy=(ax.get_xlim()[1], FLOOR), xytext=(-4, 6),
            textcoords="offset points", ha="right", color=ORANGE,
            fontsize=9, fontweight="600")

# Stagger the labels. All four sit between 192 and 242 minutes at a gain of roughly
# zero, so a single offset drew them on top of each other - four overlapping strings
# at the same height, none of them readable. Each gets its own vertical lane and a
# leader line back to its dot.
lanes = (-62, -40, -18, 20)
for c, dy in zip(sorted(under, key=lambda c: -c["minutes"])[:4], lanes):
    ax.annotate(f"{c['model']}  {c['gain_pct']:+.3f}%",
                xy=(c["minutes"], c["gain_pct"]), xytext=(-14, dy),
                textcoords="offset points", ha="right", va="center",
                fontsize=8.5, color="#52514e",
                arrowprops=dict(arrowstyle="-", color="#c9c7c0", lw=0.8,
                                shrinkA=0, shrinkB=3))

ax.set_xlabel("minutes the interval took", fontsize=9.5, color=MUTED)
ax.set_ylabel("change in validation metric (%)", fontsize=9.5, color=MUTED)
ax.legend(frameon=False, fontsize=9, loc="upper right")
ax.grid(axis="y", color="#e8e7e2", lw=1)
ax.set_axisbelow(True)
for s in ("top", "right", "left"):
    ax.spines[s].set_visible(False)
ax.spines["bottom"].set_color("#dedcd6")
ax.tick_params(colors=MUTED, labelsize=9)
plt.tight_layout()
plt.show()
```

## Per model

`reached` is where the run actually stopped; `ceiling` is the budget it was allowed. A
model marked `early` stopped because its metric flattened, not because it ran out of
budget - which is the rule doing its job.

```python
hdr = f"{'model':22}{'reached':>9}{'ceiling':>9}  {'unit':6}{'hours':>7}{'metric':>10}  early"
print(hdr)
for k, m in MODELS.items():
    met = "-" if m["metric"] is None else f"{m['metric']:.4f}"
    print(f"{k:22}{m['reached']:>9}{m['ceiling']:>9}  {m['unit']:6}"
          f"{m['hours']:>7.1f}{met:>10}  {'yes' if m['early'] else ''}")
```

## What this says, and what it does not

It says that on this hardware, with these models, **a large share of the GPU time
changed the validation metric by less than a visible amount**, and that a relative
stopping rule checked often enough would have reclaimed most of it.

It does not say these models were trained badly, or that the flat intervals were
predictable in advance - the only way to know an interval gains nothing is to measure
it. The claim is narrower: once measured, the tail should end the run.

It also does not compare architectures. The metrics come from different tasks and
different validation sets, so a PSNR here is not comparable with a PSNR there. Only
each model's change against *itself* is meaningful.
